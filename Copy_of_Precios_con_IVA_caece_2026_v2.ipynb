```python
#importaciones
import tensorflow as tf
import numpy as np
```

```python
precio = np.array([400, 500, 1000, 1500, 2000, 2500, 3000, 3500], dtype=float) #precios de entrada
precioImpuesto = np.array([484, 605, 1210, 1815, 2420, 3025, 3630, 4235], dtype=float) #precios de salida con IVA
```

```python
capa = tf.keras.layers.Dense(units=1, input_shape=[1]) #red de tipo densa, 1 capa de salida y 1 de entrada (asigno cantidad de nueronas en la capa)
modelo = tf.keras.Sequential([capa]) #modelo sequencial y le paso la capa creada
#oculta1 = tf.keras.layers.Dense(units=3, input_shape=[1]) #capa oculta de 3 nueronas, con un capa de entrada
#oculta2 = tf.keras.layers.Dense(units=3) #capa oculta de 3 nueronas
salida = tf.keras.layers.Dense(units=1) #capa de salida
#modelo = tf.keras.Sequential([oculta1, oculta2, salida])
```

```python
modelo.compile(
    optimizer=tf.keras.optimizers.Adam(0.1), #optimzador-> se pasa el factor de aprendizaje
    loss='mean_squared_error'
)
```

```python
#entrenamiento
print("Comenzando el entrenamiento")
historial = modelo.fit(precio, precioImpuesto, epochs=10, verbose=False) #paso datos de entrada, resultados esperados, epocas/cantidad veces
print("¡Modelo entrenado!")
```

```python
import matplotlib.pyplot as plt
plt.xlabel("# Epoca")
plt.ylabel("Magnitud de pérdida")
plt.plot(historial.history["loss"])
```

```python
#probemos nuestra red ya entrenada
print("¡Hagamos una predicción!")
resultado = modelo.predict(np.array([3500]))
print("El precio total del producto es " + str(resultado) + " IVA incluido!")
```
