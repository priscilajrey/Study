```python
import tensorflow as tf
import numpy as np
```

```python
precio = np.array([400, 500, 1000, 1500, 2000, 2500, 3000, 3500], dtype=float) #DATOS DE ENTRADA
precioImpuesto = np.array([484, 605, 1210, 1815, 2420, 3025, 3630, 4235], dtype=float) #SALIDA ESPERADA
```

```python
#capa = tf.keras.layers.Dense(units=1, input_shape=[1]) #ESTRUCTURA DE LA NEURONA: UNA NEURONA DE ENTRADA
#modelo = tf.keras.Sequential([capa])
#salida = tf.keras.layers.Dense(units=1) #CAPA DE SALIDA

oculta1 = tf.keras.layers.Dense(units=3, input_shape=[1]) #CAPA OCULTA
oculta2 = tf.keras.layers.Dense(units=3) #CAPA OCULTA
salida = tf.keras.layers.Dense(units=1) #CAPA DE SALIDA
modelo = tf.keras.Sequential([oculta1, oculta2, salida]) #CAPA DE SALIDA
```

```python
modelo.compile(
    optimizer=tf.keras.optimizers.Adam(0.1), #optimzador-> se pasa el factor de aprendizaje
    loss='mean_squared_error'
)
```

```python
#Entreamiento de la red
print("Comenzando el entrenamiento")
historial = modelo.fit(precio, precioImpuesto, epochs=100, verbose=False) #paso datos de entrada, resultados esperados, epocas/cantidad veces
print("¡Modelo entrenado!")
```

```python
import matplotlib.pyplot as plt
plt.xlabel("# Epoca")
plt.ylabel("Magnitud de pérdida")
plt.plot(historial.history["loss"])
```

```python
#probar
print("¡Hagamos una predicción!")
resultado = modelo.predict(np.array([500]))
print("El precio total del producto es " + str(resultado) + " IVA incluido!")
```
